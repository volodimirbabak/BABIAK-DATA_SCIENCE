```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# Завдання 1: Робота з NumPy

# 1. Створення наборів даних
arr_linear = np.arange(10, 20, 1)        
arr_random = np.random.randint(1, 50, 8) 

# 2. CRUD операції
# Create: додавання елемента
arr_linear = np.append(arr_linear, 100)

# Read: читання зрізу масиву
print("Перші 5 елементів лінійного масиву:", arr_linear[:5])

# Update: оновлення елемента за індексом
arr_random[2] = 777 

# Delete: видалення останнього елемента
arr_linear = np.delete(arr_linear, -1)

print("\n--- Результати ---")
print("Лінійний масив:", arr_linear)
print("Випадковий масив:", arr_random)
```

```python
# Завдання 2: Робота з Pandas

# 1. Створення DataFrame зі словника
data_dict = {
    'Student': ['Ivan', 'Olga', 'Petro', 'Svitlana'],
    'Math_Score': [90, 85, 72, 95],
    'City': ['Kyiv', 'Lviv', 'Dnipro', 'Odesa']
}
df_dict = pd.DataFrame(data_dict)

# 2. Створення DataFrame з масиву NumPy
data_array = np.random.randint(10, 100, (4, 3))
df_array = pd.DataFrame(data_array, columns=['Col_A', 'Col_B', 'Col_C'])

# 3. Маніпуляції з даними
print("--- Перші рядки (head) ---")
print(df_dict.head(2)) 

print("\n--- Статистичний опис (describe) ---")
print(df_array.describe())

print("\n--- Пошук даних (loc) ---")
print("Місто студента під індексом 1:", df_dict.loc[1, 'City'])

# Додавання та видалення стовпця
df_dict['Rating'] = [5, 4, 3, 5]
df_dict = df_dict.drop('Rating', axis=1)
```

```python
# Завдання 3: Візуалізація та обчислення похибок

# 1. Генерація даних для лінійної функції
m = 3.5  
b = 10   
x = np.arange(0, 10, 0.5)

y_true = m * x + b
noise = np.random.normal(0, 2.5, len(x)) # Додавання шуму
y_noisy = y_true + noise

# 2. Побудова графіка
plt.figure(figsize=(10, 6))
plt.plot(x, y_true, label=f'Теоретична функція: y={m}x+{b}', color='green', linewidth=2)
plt.scatter(x, y_noisy, label='Експериментальні дані', color='orange', s=50)

plt.title('Графік лінійної функції')
plt.xlabel('Значення X')
plt.ylabel('Значення Y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 3. Обчислення метрик похибок
MAE = np.mean(np.abs(y_true - y_noisy))
MSE = np.mean((y_true - y_noisy)**2)
print(f'MAE (Середня абсолютна похибка): {MAE:.3f}')
print(f'MSE (Середньоквадратична похибка): {MSE:.3f}')

# 4. Збереження результатів у CSV
df_results = pd.DataFrame({
    'X': x,
    'Y_noisy': y_noisy,
    'Y_hat': y_true,
    'MAE': MAE,
    'MSE': MSE
})
df_results.to_csv('Lab2_results.csv', index=False)
print("Результати успішно збережено у файл 'Lab2_results.csv'")
```
